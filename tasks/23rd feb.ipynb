```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    "Engine_Size": [1000, 1200, 1500, 1800, 2000],
    "Mileage": [25, 22, 20, 18, 15],
    "Age": [5, 4, 3, 2, 1],
    "Price": [300000, 400000, 500000, 650000, 800000]
}
df = pd.DataFrame(data)

# Features and Label
X = df[["Engine_Size", "Mileage", "Age"]]
y = df["Price"]

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict new car price
new_car = [[1500, 20, 3]]
predicted_price = model.predict(new_car)
print("Predicted Price:", predicted_price[0])

# Print coefficients
print("\nCoefficients:")
for col, val in zip(X.columns, model.coef_):
    print(col, ":", val)
```
