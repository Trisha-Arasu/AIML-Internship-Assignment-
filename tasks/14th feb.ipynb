```python
# You are given a csv file containing name, maths, science, english department
# requirement:read csv file,add new col avg, find top 3 students based on total, dept wise avg marks,students scoring above 75 in all sub, handle missing values
import pandas as pd

#read csv file
df=pd.read_csv("archive.zip")
df["total"] = df["Math"] + df["Science"] + df["English"]

# add new col avg
df["Avg"] = df["total"] / 3
print(df)
print()

# Top 3 students based on total
top3 = df.sort_values(by="total", ascending=False).head(3)
print(top3)
print()

# students scoring above 75 in all sub
df["Name"]=df["Math"]>75
df["Name"]=df["Science"]>75
df["Name"]=df["English"]>75
print(df)

#dept wise avg marks
avg_math = df["Math"].mean()
avg_science = df["Science"].mean()
avg_english = df["English"].mean()
print("Maths: ",avg_math)
print("Science: ",avg_science)
print("English: ",avg_english)

#handle missing values
df = df.fillna(0)
print(df)
```
