```python
# Mini Word2Vec Model

from gensim.models import Word2Vec

# Step 1: Create dataset (15+ sentences)
sentences = [
    "college life is fun and exciting",
    "students attend lectures daily",
    "friends study together in library",
    "exams create stress for students",
    "projects help in practical learning",
    "teachers guide students in college",
    "campus has many facilities",
    "students participate in events",
    "sports and games are important",
    "library is quiet place to study",
    "group study improves understanding",
    "assignments are part of academics",
    "students enjoy college festivals",
    "learning new skills is important",
    "college life builds future career"
]

# Step 2: Preprocess (split into words)
processed_data = [sentence.split() for sentence in sentences]

# Step 3: Train Word2Vec model
model = Word2Vec(sentences=processed_data, vector_size=50, window=3, min_count=1, workers=1)

# Step 4: Print vector of a word
word = "students"
print(f"Vector for '{word}':\n")
print(model.wv[word])

# Step 5: Find similar words
print(f"\nTop 5 words similar to '{word}':\n")
similar_words = model.wv.most_similar(word, topn=5)

for w, score in similar_words:
    print(f"{w} : {score:.3f}")
```
