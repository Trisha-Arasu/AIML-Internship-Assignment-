```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
data={
    "month":["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"],
    "revenue":[200,304,544,397,289,974,753,360,604,108,389,539],
    "marketing_spend":[345,564,234,675,934,132,122,786,743,864,457,395],
    "profit":[4,7,5,10,9,6,8,2,7,12,24,15]
}
df=pd.DataFrame(data)
```

```python
#lineplot
plt.plot(df["month"],df["revenue"],ls="--",marker="o",color="black",mfc="pink")
plt.title("Revenue trend")
plt.xlabel("Months")
plt.ylabel("Revenue")
plt.show()
```

```python
#Scatter plot
plt.scatter(df["marketing_spend"],df["profit"],alpha=1.0)
plt.title("Marketing and prfit trend")
plt.xlabel("Marketing")
plt.ylabel("Profit")
plt.show()
```

```python
#Correlation map
sns.heatmap(df[["revenue", "marketing_spend", "profit"]].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
