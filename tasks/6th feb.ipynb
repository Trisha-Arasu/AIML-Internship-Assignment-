```python
import getpass
crt_pwd = 'moni14'
attempts = 3
while attempts > 0:
    pwd = getpass.getpass("Enter the password: ")
    if pwd == crt_pwd:
        print("Access granted")
        break
    else:
        attempts -= 1
        print("There are only", attempts,"attempts left.")
if attempts == 0:
        print("Access locked!") 
```
