```python
# Smarter Mini Text Generator

import random
from collections import defaultdict

# Sample dataset (you can expand this)
sentences = [
    "college life is fun and exciting",
    "students study in library daily",
    "friends enjoy learning together",
    "teachers guide students in class",
    "projects help students learn better",
    "students participate in college events",
    "learning new skills is important",
    "college builds future career",
    "group study improves knowledge",
    "students enjoy campus life"
]

# Step 1: Build word prediction model (bigram model)
model = defaultdict(list)

for sentence in sentences:
    words = sentence.split()
    for i in range(len(words)-1):
        model[words[i]].append(words[i+1])

# Step 2: Take input from user
start_word = input("Enter a starting word: ").lower()

# Step 3: Generate 10-word sentence
result = [start_word]

for _ in range(9):  # total 10 words
    last_word = result[-1]
    
    if last_word in model:
        next_word = random.choice(model[last_word])
    else:
        next_word = random.choice(list(model.keys()))
    
    result.append(next_word)

# Step 4: Output
print("\nGenerated Sentence:")
print(" ".join(result))
```
