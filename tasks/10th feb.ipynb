```python
# Print the number is odd/even
num = int(input("Enter the number"))
if num%2==0:
    print("Even number")
else:
    print("Odd number")
```

```python
# Sum until user stops: keep asking the user to enter numbers and calculate sum.stop only when the user enter 0 - 10th feb

sum=0
while True:
    num=int(input("Enter number"))
    if num==0:
        break
    sum+=num
print(sum)
```
