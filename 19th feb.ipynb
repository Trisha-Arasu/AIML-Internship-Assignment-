```python
# Store data for 5 students using dictionary
students = {
    "A": 85,
    "B": 92,
    "C": 78,
    "D": 88,
    "E": 95
}

# Find topper
topper = max(students, key=students.get)
print("Topper:", topper, "-", students[topper])

# Calculate class average
total = sum(students.values())
average = total / len(students)
print("Class Average:", average)

# Function to assign grades
def get_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    else:
        return "D"

# Print grades
print("\nGrades:")
for name, marks in students.items():
    print(name, ":", marks, "-", get_grade(marks))
```
