```python
import string
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download required data (only first time)
nltk.download('punkt')
nltk.download('stopwords')


def clean_text(text):
    # 1. Convert to lowercase
    text = text.lower()

    # 2. Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))

    # 3. Tokenize
    words = word_tokenize(text)

    # 4. Remove stopwords
    stop_words = set(stopwords.words('english'))
    cleaned = [word for word in words if word not in stop_words]

    return cleaned


# Test cases
test_sentences = [
    "Hello!!! This is a SAMPLE text.",
    "Python is AWESOME, isn't it?",
    "I am learning NLP and it's fun :)"
]
for i, sentence in enumerate(test_sentences, 1):
    print(f"\nOriginal {i}: {sentence}")
    print("Cleaned:", clean_text(sentence))
```
