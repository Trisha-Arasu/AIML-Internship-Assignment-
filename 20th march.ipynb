```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('stopwords')

# 20 messy sentences
sentences = [
    "Heyyy brooo!!! 😂😂 what's up??",
    "Win $$$ now!!! click here 👉",
    "Gud mrng frnd 😊 hv a nice day",
    "LOL that was sooo funny 🤣",
    "Ur account has been hacked!!!",
    "Free gift waiting 4 u 🎁",
    "C u soon!!!",
    "This is amaznggg 😍",
    "Call me asap plzzz",
    "OMG!!! unbelievable 😱",
    "Get rich quick $$$",
    "Hiiiii how r uuu?",
    "Dont miss this offer!!!",
    "Best deal everrrrr",
    "Congrats!!! u won prize 🏆",
    "Why u late???",
    "I luv this ❤️",
    "Pls send details asap",
    "Thnks a lot 😊",
    "Join now & earn money!!!"
]

# slang dictionary
slang_dict = {
    "u": "you", "ur": "your", "r": "are", "hv": "have",
    "frnd": "friend", "pls": "please", "luv": "love",
    "asap": "as soon as possible", "lol": "laughing"
}

# emoji pattern
emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"
                           u"\U0001F300-\U0001F5FF"
                           u"\U0001F680-\U0001F6FF"
                           u"\U0001F1E0-\U0001F1FF"
                           "]+", flags=re.UNICODE)

stop_words = set(stopwords.words('english'))

for i, text in enumerate(sentences, 1):
    print(f"\nSentence {i}: {text}")

    # detect emojis
    emojis = emoji_pattern.findall(text)

    # lowercase
    text = text.lower()

    # remove emojis
    text = emoji_pattern.sub(r'', text)

    # replace slang
    words = text.split()
    words = [slang_dict.get(word, word) for word in words]

    text = " ".join(words)

    # remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # tokenize
    tokens = word_tokenize(text)

    # remove stopwords
    filtered = [w for w in tokens if w not in stop_words]

    print("Emojis found:", emojis)
    print("Cleaned:", filtered)
```
