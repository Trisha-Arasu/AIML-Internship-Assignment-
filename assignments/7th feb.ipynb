```python
import re
import hashlib

# Function to check strong password
def is_strong_password(password):
    if len(password) < 8:
        return False
    if not re.search(r"[A-Z]", password):
        return False
    if not re.search(r"[a-z]", password):
        return False
    if not re.search(r"[0-9]", password):
        return False
    if not re.search(r"[!@#$%^&*]", password):
        return False
    return True

# Function to hash password
def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()

# Input from user
stored_password = input("Create a password: ")

# Check strength
if is_strong_password(stored_password):
    stored_hash = hash_password(stored_password)
    print("Password is strong and stored securely.")
else:
    print("Weak password. Try again.")
    exit()

# Authentication
entered_password = input("Enter password to login: ")

if hash_password(entered_password) == stored_hash:
    print("Authentication successful!")
else:
    print("Authentication failed!")
```
