```python
import pandas as pd

# Create dataset
data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [40, 50, 60, 70, 80]
}
df = pd.DataFrame(data)
print("Dataset:")
print(df)

# Features and Labels
X = df[["Study_Hours"]]   # Feature
y = df["Marks"]           # Label
print("\nFeature (X):")
print(X)
print("\nLabel (y):")
print(y)
```
