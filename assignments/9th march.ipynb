```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Create dataset
data = {
    "Area": [500, 1000, 1500, 2000, 2500],
    "Price": [1000000, 2000000, 3000000, 4000000, 5000000]
}
df = pd.DataFrame(data)

# Features and Label
X = df[["Area"]]   # Feature
y = df["Price"]    # Label

# Train model
model = LinearRegression()
model.fit(X, y)

# Predict price for new input
new_area = [[1800]]
predicted_price = model.predict(new_area)
print("Predicted price for 1800 sq.ft:", predicted_price[0])
```
